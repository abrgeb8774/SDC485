```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import simpy
from datetime import datetime
from IPython.display import display, Markdown
todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "abrgeb8774"
display(Markdown("### Import Necessary Libraries"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
```

```python
display(Markdown("### Define Scenario Parameters"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Define scenario parameters
principal = 300000
interest_rate_scenarios = [0.03, 0.05, 0.08]
loan_term_scenarios = [15, 30, 40]
monthly_payment_scenarios = []
# Calculate monthly loan payments
scenario_number = 0
for interest_rate in interest_rate_scenarios:
    loan_term = loan_term_scenarios[scenario_number]
    r = interest_rate / 12
    n = loan_term * 12
    M = principal * r * (1 + r)**n / ((1 + r)**n - 1)
    monthly_payment_scenarios.append(M)
    scenario_number = scenario_number + 1
# Display scenario parameters
print("Principal Amount: $", principal)
print("Interest Rate Scenarios:", interest_rate_scenarios)
print("Loan Term Scenarios (Years):", loan_term_scenarios)
print("Monthly Payment Scenarios:", monthly_payment_scenarios)
```

```python
display(Markdown("### Perform Scenario Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Define function to calculate monthly loan repayment
def calculate_monthly_payment(principal, annual_interest_rate, loan_term):
    r = annual_interest_rate / 12
    n = loan_term * 12
    M = principal * r * (1 + r)**n / ((1 + r)**n - 1)
    return M
# Perform scenario analysis
results = []
for i in range(len(interest_rate_scenarios)):
    interest_rate = interest_rate_scenarios[i]
    loan_term = loan_term_scenarios[i]
    M = calculate_monthly_payment(principal, interest_rate, loan_term)
    total_repayment = M * loan_term * 12
    results.append({
        "Scenario": f"Scenario {i + 1}",
        "Interest_Rate": interest_rate,
        "Loan_Term_Years": loan_term,
        "Monthly_Payment": M,
        "Total_Repayment": total_repayment
    })
# Convert results to DataFrame
results_df = pd.DataFrame(results)
# Display results
display(results_df)
```

```python
display(Markdown("### Visualize Total Loan Repayment"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Plot the total loan repayment amounts across scenarios
plt.figure(figsize=(12, 6))
sns.barplot(data=results_df, x="Scenario", y="Total_Repayment")
plt.title("Comparison of Total Loan Repayment Across Scenarios")
plt.xlabel("Loan Repayment Scenarios")
plt.ylabel("Total Loan Repayment Amount ($)")
plt.grid(True)
plt.show()
```

```python
display(Markdown("### Apply Sensitivity Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
# Modify scenario parameters
modified_interest_rate_scenarios = [0.04, 0.06, 0.09]
# Perform sensitivity analysis
sensitivity_results = []
scenario_number = 0
for interest_rate in modified_interest_rate_scenarios:
    loan_term = loan_term_scenarios[scenario_number]
    r = interest_rate / 12
    n = loan_term * 12
    M = principal * r * (1 + r)**n / ((1 + r)**n - 1)
    total_repayment = M * n
    sensitivity_results.append({
        "Scenario": f"Scenario {scenario_number + 1}",
        "Interest_Rate": interest_rate,
        "Total_Repayment": total_repayment
    })
    scenario_number = scenario_number + 1
# Convert results to DataFrame
sensitivity_df = pd.DataFrame(sensitivity_results)
# Display results
display(sensitivity_df)
# Plot the sensitivity analysis results
plt.figure(figsize=(12, 6))
sns.barplot(data=sensitivity_df, x="Scenario", y="Total_Repayment")
plt.title("Total Loan Repayment Under Modified Interest Rates")
plt.xlabel("Loan Repayment Scenarios")
plt.ylabel("Total Loan Repayment Amount ($)")
plt.grid(True)
plt.show()
```
